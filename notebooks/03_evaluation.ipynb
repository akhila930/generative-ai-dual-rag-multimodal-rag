# 03 — RAG Evaluation

We use a small ground-truth dataset created directly from the handbook.

The notebook computes practical proxies for:
- context precision
- context recall
- answer correctness
- answer relevancy
- faithfulness
- latency

The proxies are transparent and reproducible. For the final report, describe them as local evaluation proxies. If a fully model-based RAGAS run is feasible on your machine, add it as an extension.

```python
from src.evaluation import run_evaluation

smol_rows = run_evaluation("smollm")
llama_rows = run_evaluation("llama")
print("SmolLM rows:", len(smol_rows))
print("Llama rows:", len(llama_rows))
```

```python
import pandas as pd
from pathlib import Path

results_dir = Path("results")
smol_df = pd.read_csv(results_dir / "smollm_evaluation.csv")
llama_df = pd.read_csv(results_dir / "llama_evaluation.csv")

metric_cols = [
    "context_precision_proxy",
    "context_recall_proxy",
    "answer_correctness_proxy",
    "answer_relevancy_proxy",
    "faithfulness_proxy",
    "latency_seconds",
]

summary = pd.concat([smol_df, llama_df]).groupby("model")[metric_cols].mean()
summary
```
